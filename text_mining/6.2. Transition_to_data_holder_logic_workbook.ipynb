unique_institutions = (
    df['standardized_data_holder_name']
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
    .drop_duplicates()
)

unique_institutions = unique_institutions.to_frame("data_holder")

unique_institutions["holder_ISO2_country_new"] = ""
unique_institutions["holder_type_new"] = ""
unique_institutions["notes"] = ""

with pd.ExcelWriter(
    file,
    engine="openpyxl",
    mode="a",
    if_sheet_exists="replace"
    ) as writer:
        unique_institutions.to_excel(
            writer,
            sheet_name="unique_institutions",
            index=False
        )
